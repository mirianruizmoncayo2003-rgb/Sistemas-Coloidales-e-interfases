```python

import numpy as np
from scipy.special import exp1
from scipy.integrate import quad

def henry(alpha):

    if 0 < alpha < 5:
    
        I = exp1(alpha)

        f = (1 + alpha**2/16- 5*alpha**3/48- alpha**4/96+ alpha**5/96+ (alpha**4/8 - alpha**6/96)* np.exp(alpha) * I)

    elif alpha > 25:
        
        f = (3/2 - 9/(2*alpha)+ 75/(2*alpha**2)- 330/alpha**3)

    else:

        f_4=(1 + 4**2/16- 5*4**3/48- 4**4/96+ 4**5/96+ (4**4/8 - 4**6/96)* np.exp(4) * exp1(4))
        
        f_30=(3/2 - 9/(2*30)+ 75/(2*30**2)- 330/30**3)

        f= f_4 + (f_30 - f_4) * (np.log10(alpha) - np.log10(4)) / (np.log10(30) - np.log10(4))


    return f


def expn(n, alpha):
    integral, _ = quad(lambda u: np.exp(-u) / (1 + u/alpha)**n,0, np.inf)
    return integral / alpha

def henry_funcion_E(alpha):
        E7 = expn(7, alpha)
        E5 = expn(5, alpha)
        f_1 = 1.5 * (1 - 5*E7 + 2*E5)
        return f_1

valores = [254.21, 2.32, 0.9, 56.84, 5.68, 7, 10, 13, 15, 17, 20, 23]

for alpha in valores:
    print(f"α = {alpha:5.1f}  f(α) = {henry(alpha):.6f}  f_r(α) = {henry_funcion_E(alpha):.6f}" )
   
  
```
